```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
data = titanic['Survived'].value_counts()
plt.pie(data,labels=data.index,colors=['red','blue'],
       autopct='%0.2f%%',wedgeprops = {'width':0.5})
plt.title('Servival')
plt.show()
```

# 2. Count plot

```python
sns.countplot(data = iris,x='species' )
plt.title('Species')
plt.show()
```

# 3. Kde + Rug plot

```python
sns.kdeplot(tips['total_bill'])
sns.rugplot(tips['total_bill'])
plt.title('Total Bill')
plt.show()
```

# 4. Violin plot

```python
sns.violinplot(data = tips['tip'],color = 'blue')
plt.title('This graph shows tip')
plt.show()
```

# 5. Box plot

```python
sns.boxplot(data = titanic, x='Fare')
plt.title('Box plot on Fare of passengers')
```

# 6. Boxen plot

```python
sns.boxenplot(titanic['Fare'])
plt.title('Boxen plot on fare row')
plt.show()
```

# 7. Histogram

```python
sns.histplot(iris['sepal_width'])
plt.title('sepal_width of all iris')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = tips,x='total_bill',y='tip')
plt.title('Scatter plot on total_bill and tip')
plt.show()
```

# 2. Line plot

```python
cnt = titanic['Age'].value_counts()
sns.lineplot(x = cnt.index, y = cnt.values)
plt.title('Age')
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(data= iris,x='sepal_length',y='sepal_width')
plt.title('Joint plot between sepal length and sepal width')
plt.show()
```

# 4. Hexbin plot

```python
plt.hexbin(iris['sepal_length'],iris['sepal_width'],gridsize=28,cmap='Blues')
plt.title('Hexbin Plot: Sepal Length vs Sepal Width')
plt.colorbar(label="count in bin")
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

# 5. 2D kde plot

```python
sns.kdeplot(data = tips, x='total_bill',y = 'tip',fill = True)
plt.title('2D KDE Plot between Total Bill and tip')
plt.xlabel='Total_Bill'
plt.ylabe='Tip'
plt.show()
```

# 6. Box plot

```python
sns.boxplot(data = titanic , x='Survived',y='Age',palette = 'Set1')
plt.title('Boxplot on survived according to age')
plt.show()
```

# 7. Violin plot

```python
sns.violinplot(data = titanic ,x = 'Survived', y = 'Pclass')
plt.show()
```

# 8. Swarm plot

```python
sns.swarmplot(data = tips,x='tip',y='day',hue='day')
plt.title('Swarm plot between Tip and day of Tips dataset')
plt.show()
```

# 9. Count plot

```python
sns.countplot(data = titanic, x = 'Pclass',hue='Survived' )
plt.title('Count plot on pclass and survived passenger')
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = iris, x = 'sepal_length', y = 'sepal_width', hue = 'species')
plt.title('Scatter plot on sepal length and sepal width to find species')
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(data = tips , hue='time')
```

# 3. Heatmap

```python
sns.heatmap(data = tips.corr(numeric_only=True))
plt.show()
```

# 4. FacetGrid plot

```python
D = sns.FacetGrid(tips,col = 'day',row = 'time')
D.map(sns.scatterplot, 'total_bill','tip')
plt.show()
```

# 5. PairGrid

```python
E = sns.PairGrid(tips)
E.map(sns.scatterplot)
plt.show()
```

# 6. Joint plot

```python

sns.jointplot(data = tips, x = 'total_bill', y = 'tip', hue = 'sex')
plt.show()
```

# 7. Boxen plot

```python

sns.boxenplot(data = titanic, x='Pclass', y='Fare', hue='Survived')

plt.title('Boxen plot on Fare Distribution by Passenger Pclass and Survived')
plt.show()
```
